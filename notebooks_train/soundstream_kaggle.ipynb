```python
import comet_ml

api = comet_ml.API(api_key=user_secrets.get_secret("COMET_API_KEY"))
exp = api.get_experiment("german-zverev", "soundstream", "5xkp7muwkj0gp01akaqfblrwqs29pakr")

assets = exp.get_asset_list()
model_asset = [a for a in assets if "model_best" in a['fileName']]

asset_id = model_asset[0]['assetId']
asset_binary = exp.get_asset(asset_id)

with open("./model_best.pth", "wb") as f:
    f.write(asset_binary)
```

```python
!python train.py \
    --config-name=soundstream_gan \
    trainer.save_dir="/kaggle/working/saved" \
    +datasets.train.data_dir="/kaggle/input/datasets/a24998667/librispeech" \
    +datasets.test.data_dir="/kaggle/input/datasets/a24998667/librispeech" \
    writer.log_checkpoints=True \
    +writer.run_id="5xkp7muwkj0gp01akaqfblrwqs29pakr" \
    trainer.resume_from="./model_best.pth"
```
